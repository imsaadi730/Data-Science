# Decision Trees

A **Decision Tree** predicts an outcome by asking a sequence of **if-then questions** about the input features.

A Decision Tree works like a tree structure where data is divided based on different conditions.

---

## Decision Tree Algorithms

### 1. CART

**CART** stands for:

**Classification and Regression Trees**

---

### 2. ID3 / C4.5

**ID3** and **C4.5** are Decision Tree algorithms used for making splits in the data.

---

### 3. CHAID

**CHAID** stands for:

**Chi-squared Automatic Interaction Detector**

It uses the **Chi-Square** concept for making decisions and splitting the data.

---

## Important Concepts

### Entropy

**Entropy** measures the impurity or randomness in the data.

- Lower Entropy → More pure data
- Higher Entropy → More impure data

---

### Gini Impurity

**Gini Impurity** is another measure used to determine the impurity of a node.

It is commonly used in **CART**.

---

## Decision Tree Prediction

A Decision Tree asks questions about the input features step-by-step.

For example:

```text
Feature?
   |
   ├── Yes → Class 1
   |
   └── No  → Class 0
