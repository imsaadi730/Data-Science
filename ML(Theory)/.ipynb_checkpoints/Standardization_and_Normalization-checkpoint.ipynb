# Feature Scaling (Standardization)

## What is Feature Scaling?

Feature Scaling is the process of bringing all numerical features to a similar scale.

It helps Machine Learning algorithms learn faster and perform better.

---

# Why Do We Need Feature Scaling?

Suppose we have the following features:

| Age | Salary |
|----:|-------:|
| 25 | 25,000 |
| 35 | 80,000 |
| 45 | 120,000 |

Here,

- Age ranges between **20–60**
- Salary ranges between **20,000–120,000**

Since Salary has much larger values, it can dominate the learning process.

Feature Scaling makes both features comparable.

---

# Standardization

Standardization transforms data so that:

- Mean = 0
- Standard Deviation = 1

Formula:

```text
Z = (X - Mean) / Standard Deviation
```

where

- X = Original Value
- Mean = Average of the feature
- Standard Deviation = Spread of the feature

---

# Example

Suppose Age values are:

```text
20, 30, 40, 50, 60
```

Mean:

```text
40
```

Standard Deviation:

```text
14.14
```

Now Standardize:

| Original | Standardized |
|----------:|-------------:|
| 20 | -1.41 |
| 30 | -0.71 |
| 40 | 0 |
| 50 | 0.71 |
| 60 | 1.41 |

Now the feature has

- Mean ≈ 0
- Standard Deviation ≈ 1

---

# Why Fit on Training Data Only?

We calculate the Mean and Standard Deviation **only from the training data**.

Reason:

To prevent **Data Leakage**.

The test data should remain completely unseen during training.

---

---

# Difference Between fit(), transform() and fit_transform()

### fit()

- Calculates Mean
- Calculates Standard Deviation
- Does NOT change the data

Example:

```python
scaler.fit(X_train)
```

---

### transform()

Uses the previously calculated Mean and Standard Deviation to scale the data.

Example:

```python
X_train = scaler.transform(X_train)
```

---

### fit_transform()

Performs both operations together.

Equivalent to:

```python
scaler.fit(X_train)
X_train = scaler.transform(X_train)
```

---

# Why Don't We Use fit_transform() on Test Data?

Incorrect:

```python
X_test = scaler.fit_transform(X_test)
```

This calculates a new Mean and Standard Deviation from the test data.

The model would see information from the test set, causing **Data Leakage**.

Correct:

```python
X_test = scaler.transform(X_test)
```

---

# Complete Code

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

---

# Algorithms That Need Scaling

Scaling is recommended for:

- Logistic Regression
- Linear Regression
- KNN
- SVM
- K-Means Clustering
- PCA
- Neural Networks

---

# Algorithms That Do NOT Need Scaling

Tree-based algorithms generally do not require scaling.

Examples:

- Decision Tree
- Random Forest
- XGBoost
- LightGBM
- CatBoost

---

# Key Points

- Standardization makes features have Mean = 0 and Standard Deviation = 1.
- Always use `fit_transform()` on the training data.
- Always use `transform()` on the test data.
- Never use `fit_transform()` on the test data.
- Standardization improves the performance of many Machine Learning algorithms.
