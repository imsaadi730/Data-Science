# k-Nearest Neighbours (kNN)

## Introduction

**k-Nearest Neighbours (kNN)** is a **Supervised Machine Learning Algorithm** that is used for both **Classification** and **Regression** problems.

It is called a **Lazy Learning Algorithm** because it does **not build any model during the training phase**. Instead, it stores the complete training dataset and performs all calculations only when a new data point is given.

---

# Working of kNN Algorithm

The kNN algorithm follows the following steps.

## Step 1: Choose the Value of K

- Choose the number of nearest neighbours (**K**).
- **K** is called a **Hyperparameter**.
- In Classification problems, an **odd value of K** (3, 5, 7, ...) is usually preferred to avoid ties.

### Example

If

$$
K = 5
$$

then the algorithm will consider the **5 nearest neighbours** while making the prediction.

---

## Step 2: Calculate the Distance

For the given **new data point**, calculate its distance from every training sample.

After calculating all distances:

1. Sort the distances in ascending order.
2. Select the **K nearest neighbours** having the smallest distances.

The prediction depends on the selected distance metric.

---

# Distance Metrics

## 1. Euclidean Distance

The Euclidean Distance is the **shortest straight-line distance** between two points.

Suppose there are two points

$$
P_1(x_1,y_1)
$$

and

$$
P_2(x_2,y_2)
$$

### Formula

$$
d(P_1,P_2)=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

### Step-by-Step Calculation

Difference along x-axis

$$
\Delta x=x_2-x_1
$$

Difference along y-axis

$$
\Delta y=y_2-y_1
$$

Square both differences

$$
(\Delta x)^2=(x_2-x_1)^2
$$

$$
(\Delta y)^2=(y_2-y_1)^2
$$

Add both squared values

$$
(x_2-x_1)^2+(y_2-y_1)^2
$$

Finally, take the square root

$$
d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

### Example

Suppose

$$
P_1=(2,3)
$$

$$
P_2=(5,7)
$$

Then

$$
d=\sqrt{(5-2)^2+(7-3)^2}
$$

$$
=\sqrt{3^2+4^2}
$$

$$
=\sqrt{9+16}
$$

$$
=\sqrt{25}
$$

$$
d=5
$$

---

## 2. Manhattan Distance

The Manhattan Distance is also called the **City Block Distance** because it measures distance by moving horizontally and vertically.

### Formula

$$
d(P_1,P_2)=|x_2-x_1|+|y_2-y_1|
$$

### Step-by-Step Calculation

Difference along x-axis

$$
|x_2-x_1|
$$

Difference along y-axis

$$
|y_2-y_1|
$$

Add both values

$$
d=|x_2-x_1|+|y_2-y_1|
$$

### Example

Suppose

$$
P_1=(2,3)
$$

$$
P_2=(5,7)
$$

Then

$$
d=|5-2|+|7-3|
$$

$$
=3+4
$$

$$
d=7
$$

---

# Step 3: Make Prediction

After selecting the **K nearest neighbours**, the prediction is made as follows.

## Classification

For Classification:

- Find the class label of each neighbour.
- Count how many neighbours belong to each class.
- Select the class having the **highest frequency (Majority Voting).**

### Example

Suppose

$$
K=5
$$

Nearest neighbours are

| Neighbour | Class |
|-----------|-------|
| 1 | A |
| 2 | A |
| 3 | B |
| 4 | A |
| 5 | B |

Number of neighbours

$$
A=3
$$

$$
B=2
$$

Since

$$
3>2
$$

Therefore

$$
\boxed{\text{Predicted Class = A}}
$$

---

## Regression

For Regression:

Take the **average (mean)** of the target values of the K nearest neighbours.

### Formula

$$
\hat{y}=\frac{1}{K}\sum_{i=1}^{K}y_i
$$

or

$$
\hat{y}=\frac{y_1+y_2+y_3+\cdots+y_K}{K}
$$

### Example

Suppose

$$
K=5
$$

Nearest neighbour values are

$$
20,\;22,\;24,\;26,\;28
$$

Then

$$
\hat{y}=\frac{20+22+24+26+28}{5}
$$

$$
=\frac{120}{5}
$$

$$
=24
$$

Therefore

$$
\boxed{\hat{y}=24}
$$

---

# Comparison of Distance Metrics

| Feature | Euclidean Distance | Manhattan Distance |
|---------|---------------------|--------------------|
| Formula | $\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}$ | $|x_2-x_1|+|y_2-y_1|$ |
| Path | Straight Line | Horizontal & Vertical |
| Best For | Continuous Numerical Data | Grid-based & High-dimensional Data |
| Sensitive to Outliers | Yes | Less |

---

# Advantages of kNN

- Simple and easy to understand.
- Easy to implement.
- No training phase is required.
- Works for both Classification and Regression.
- Can model complex decision boundaries.
- Effective for small datasets.

---

# Disadvantages of kNN

- Slow for large datasets.
- Requires storing the entire training dataset.
- Sensitive to irrelevant features.
- Performance depends on the value of **K**.
- Feature Scaling is required.
- Computationally expensive during prediction.

---

# Summary

The complete workflow of the **k-Nearest Neighbours (kNN)** algorithm is:

1. Choose the value of **K**.
2. Calculate the distance between the new data point and all training samples.
3. Select the **K nearest neighbours**.
4. For **Classification**, use **Majority Voting**.
5. For **Regression**, calculate the **Mean (Average)** of the neighbours' target values.

> **kNN is one of the simplest and most powerful supervised machine learning algorithms. It is an excellent algorithm for beginners because it is easy to understand and forms the foundation for many other machine learning concepts.**

# ⚠️ Limitations of k-Nearest Neighbours (kNN)

Although **k-Nearest Neighbours (kNN)** is one of the simplest and most effective Machine Learning algorithms, it has several limitations that can affect its performance, especially when working with large or high-dimensional datasets.

---

# 1. Slow During Prediction

One of the biggest disadvantages of kNN is that it is **slow at prediction time**.

### Why?

When a new data point is given, the algorithm must:

1. Calculate the distance from the new point to **every training sample**.
2. Sort all the calculated distances.
3. Find the **K nearest neighbours**.
4. Make the prediction.

Since these calculations are performed **every time** a new sample is predicted, the prediction process becomes slow.

### Problem

- Not suitable for **large datasets**.
- Prediction time increases as the dataset size increases.

---

# 2. Memory Intensive

The kNN algorithm stores the **entire training dataset** because it does not learn any mathematical model during training.

Therefore,

- More data requires more memory.
- Memory usage increases continuously as the dataset grows.

Hence, kNN is considered a **memory-intensive algorithm**.

---

# 3. Sensitive to Outliers

An **Outlier** is a data point that is very far away from the rest of the data.

If an outlier becomes one of the nearest neighbours, it may influence the prediction and reduce the accuracy of the model.

### Example

Suppose most points belong to **Class A**, but one abnormal point from **Class B** lies very close to the new sample.

In this case, the prediction may become incorrect because of that outlier.

---

# 4. Sensitive to Feature Scaling

kNN calculates the distance between data points.

If different features have different ranges (magnitudes), then the feature with the larger values dominates the distance calculation.

### Example

Suppose a dataset contains two features:

- Age = 25
- Salary = 500000

Since **Salary** has much larger values than **Age**, the distance calculation will be mostly influenced by Salary.

As a result,

- Age contributes very little.
- Predictions become biased.

### Solution

Before applying kNN, perform **Feature Scaling** using techniques such as:

- Standardization (Z-score Normalization)
- Min-Max Scaling

Feature scaling ensures that all features contribute equally during distance calculation.

---

# 5. Curse of Dimensionality

The **Curse of Dimensionality** occurs when the number of features (dimensions) becomes very large.

### High-Dimensional Data

Suppose a dataset has:

- 2 Features → Easy to calculate distances
- 5 Features → Slightly difficult
- 100 Features → Much more difficult
- 1000 Features → Very difficult

As the number of dimensions increases:

- Data points become increasingly sparse.
- The distance between points becomes less meaningful.
- Nearly all points appear to be at similar distances.

This phenomenon is called **Data Sparsity**.

Because of this,

- Finding true nearest neighbours becomes difficult.
- Model accuracy decreases.
- kNN performs poorly on high-dimensional datasets.

---

# 📌 Summary of Limitations

| Limitation | Description |
|------------|-------------|
| **Slow Prediction** | Distance is calculated from every training sample during prediction. |
| **Memory Intensive** | Entire training dataset must be stored in memory. |
| **Sensitive to Outliers** | Outliers can significantly affect predictions. |
| **Sensitive to Feature Scaling** | Features with larger values dominate distance calculations. |
| **Curse of Dimensionality** | Performance decreases as the number of features increases due to data sparsity. |

---

# ✅ Key Takeaways

- kNN has **no training phase**, but its **prediction phase is computationally expensive**.
- It requires storing the **complete training dataset**, making it memory-intensive.
- **Feature Scaling** is essential before applying kNN.
- Outliers should be handled during data preprocessing.
- kNN is generally **not recommended for very large or high-dimensional datasets** because of the **Curse of Dimensionality**.

````python
import pandas as pd
from sklearn.metrics import precision_score,accuracy_score,recall_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
````

````python
df = pd.read_csv("heart.csv")
````

````python
df.head()
````

````python
x=df.drop("target" ,axis=1)
y=df["target"]
````

````python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
````

````python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
````

````python
model=KNeighborsClassifier(n_neighbors=3)
model.fit(x_train_scaled,y_train)
````

````python
y_pred=model.predict(x_test_scaled)
````

````python
print("Recall Score :",recall_score(y_test,y_pred))
print("Accuracy_score:",accuracy_score(y_test,y_pred))
print("Precision score :",precision_score(y_test,y_pred))
````

````python
# For n_neighbors=7
model=KNeighborsClassifier(n_neighbors=7)
model.fit(x_train_scaled,y_train)
y_pred=model.predict(x_test_scaled)
print("Recall Score :",recall_score(y_test,y_pred))
print("Accuracy_score:",accuracy_score(y_test,y_pred))
print("Precision score :",precision_score(y_test,y_pred))
````

# 📘 Validation Data

## Introduction

In Machine Learning, after preparing the dataset, it is **not a good practice** to train the model using the entire dataset. Instead, the dataset is divided into different parts so that the model can be trained, tuned, and evaluated properly.

A **Validation Dataset** is a portion of the training data that is **held back during training** and is used to **tune the model's hyperparameters** and estimate its performance before testing on unseen data.

---

# Dataset Splitting

A dataset is generally divided into two major parts:

- **Training Data**
- **Testing Data**

A common split is:

- **Training Data = 80%**
- **Testing Data = 20%**

```
Complete Dataset
│
├── Training Data (80%)
└── Testing Data (20%)
```

However, when hyperparameter tuning is required, the **Training Data** is further divided into:

- **Training Set**
- **Validation Set**

---

# Training, Validation, and Testing Split

A commonly used split is:

- **Training Set = 60%**
- **Validation Set = 20%**
- **Testing Set = 20%**

```
Complete Dataset
│
├── Training Set (60%)
├── Validation Set (20%)
└── Testing Set (20%)
```

---

# Purpose of Each Dataset

## 1. Training Set

The **Training Set** is used to train the Machine Learning model.

During this phase:

- The model learns patterns from the data.
- Model parameters (weights) are updated.
- No performance evaluation is done on unseen data.

---

## 2. Validation Set

The **Validation Set** is used during the training process to evaluate different versions of the model.

It helps in:

- Hyperparameter Tuning
- Model Selection
- Preventing Overfitting
- Comparing different models

The validation data is **never used to train the model**.

Instead, it is used only to check how well the trained model performs.

---

## 3. Testing Set

The **Testing Set** is used only after the final model has been selected.

Its purpose is to measure the model's performance on completely unseen data.

The testing data should **never be used** during training or hyperparameter tuning.

---

# Hyperparameter Tuning using Validation Data

Some Machine Learning algorithms require choosing hyperparameters before training.

For example, in **k-Nearest Neighbours (kNN)**,

the hyperparameter is

\[
K
\]

Different values of **K** are tested, such as

\[
K = 1,\;3,\;5,\;7,\;9,\;11
\]

The process is:

1. Train the model using the **Training Set**.
2. Evaluate the model using the **Validation Set**.
3. Compare the performance for different values of **K**.
4. Select the value of **K** that gives the highest validation accuracy.
5. Finally, evaluate the selected model using the **Testing Set**.

---

# Workflow

```
Training Set
      │
      ▼
Train the Model
      │
      ▼
Validation Set
      │
      ▼
Tune Hyperparameters
      │
      ▼
Select Best Model
      │
      ▼
Testing Set
      │
      ▼
Final Performance
```

---

# Example

Suppose we are using **kNN**.

We try different values of **K**.

| Value of K | Validation Accuracy |
|------------|--------------------:|
| 1 | 88% |
| 3 | 91% |
| 5 | 94% |
| 7 | 92% |
| 9 | 90% |

The highest validation accuracy is obtained when

\[
K = 5
\]

Therefore,

\[
\boxed{\text{Final Hyperparameter } K = 5}
\]

Now, this final model is evaluated on the **Testing Set**.

---

# Why Do We Need Validation Data?

Without a validation set,

- We cannot choose the best hyperparameters.
- The model may become overfitted.
- The testing data may accidentally influence the model.

Using a validation dataset helps us build a model that generalizes better to unseen data.

---

# Difference Between Training, Validation, and Testing Data

| Dataset | Purpose | Used for Training? | Used for Hyperparameter Tuning? |
|---------|---------|-------------------|---------------------------------|
| **Training Set** | Learn patterns from data | ✅ Yes | ❌ No |
| **Validation Set** | Tune hyperparameters & select the best model | ❌ No | ✅ Yes |
| **Testing Set** | Evaluate the final model | ❌ No | ❌ No |

---

# Key Points

- The **Training Set** is used to train the model.
- The **Validation Set** is used for **Hyperparameter Tuning** and **Model Selection**.
- The **Testing Set** is used only once to evaluate the final model.
- A common split is **60% Training, 20% Validation, and 20% Testing**.
- In **kNN**, the Validation Set helps determine the best value of **K** before testing the final model.

---

# Summary

The complete workflow is:

1. Split the dataset.
2. Train the model using the **Training Set**.
3. Evaluate different hyperparameters using the **Validation Set**.
4. Select the best-performing model.
5. Test the final model on the **Testing Set** to obtain the final performance.

> **The Testing Set should never be used during training or hyperparameter tuning because it represents unseen data and provides an unbiased estimate of the model's real-world performance.**

# 📘 Cross Validation (K-Fold Cross Validation)

## Introduction

When building a Machine Learning model, we want to know how well it will perform on **unseen data**. If we evaluate the model using only one train-validation split, the result may depend heavily on that particular split.

To obtain a **more reliable and unbiased estimate** of model performance, we use **Cross Validation**.

One of the most popular techniques is **K-Fold Cross Validation**.

---

# What is K-Fold Cross Validation?

In **K-Fold Cross Validation**, the training dataset is divided into **K equal-sized subsets (folds)**.

- One fold is used as the **Validation Set**.
- The remaining **K − 1 folds** are used as the **Training Set**.
- This process is repeated **K times**, so that every fold is used **exactly once** as the validation set.

Finally, the performance of all K models is averaged to obtain a more reliable estimate.

---

# Example

Suppose we have a dataset containing

\[
1000 \text{ samples}
\]

We first split the dataset into:

- **Training Data = 800 samples (80%)**
- **Testing Data = 200 samples (20%)**

The testing data is kept aside and is **never used** during Cross Validation.

Now suppose

\[
K = 5
\]

The **800 training samples** are divided into **5 equal folds**.

Since

\[
\frac{800}{5}=160
\]

each fold contains

\[
160 \text{ samples}
\]

---

# Working of 5-Fold Cross Validation

### Iteration 1

```
Validation   Training   Training   Training   Training
 Fold 1       Fold 2      Fold 3      Fold 4      Fold 5
```

- Train the model on **Fold 2 + Fold 3 + Fold 4 + Fold 5**
- Validate the model on **Fold 1**
- Store **Validation Accuracy 1**

---

### Iteration 2

```
Training   Validation   Training   Training   Training
 Fold 1      Fold 2       Fold 3      Fold 4      Fold 5
```

- Train on **Fold 1 + Fold 3 + Fold 4 + Fold 5**
- Validate on **Fold 2**
- Store **Validation Accuracy 2**

---

### Iteration 3

```
Training   Training   Validation   Training   Training
 Fold 1      Fold 2      Fold 3       Fold 4      Fold 5
```

- Train on all folds except **Fold 3**
- Validate on **Fold 3**
- Store **Validation Accuracy 3**

---

### Iteration 4

```
Training   Training   Training   Validation   Training
 Fold 1      Fold 2      Fold 3      Fold 4       Fold 5
```

- Train on all folds except **Fold 4**
- Validate on **Fold 4**
- Store **Validation Accuracy 4**

---

### Iteration 5

```
Training   Training   Training   Training   Validation
 Fold 1      Fold 2      Fold 3      Fold 4      Fold 5
```

- Train on all folds except **Fold 5**
- Validate on **Fold 5**
- Store **Validation Accuracy 5**

---

# Final Validation Score

Suppose the validation accuracies are

| Iteration | Validation Accuracy |
|-----------|--------------------:|
| 1 | 91% |
| 2 | 93% |
| 3 | 92% |
| 4 | 94% |
| 5 | 90% |

The final Cross Validation Accuracy is calculated by taking the average of all validation accuracies.

\[
\boxed{
\text{Cross Validation Accuracy}
=
\frac{A_1+A_2+A_3+A_4+A_5}{5}
}
\]

Substituting the values,

\[
=
\frac{91+93+92+94+90}{5}
\]

\[
=
\frac{460}{5}
\]

\[
=
92\%
\]

Therefore,

\[
\boxed{\text{Final Cross Validation Accuracy}=92\%}
\]

---

# Hyperparameter Tuning using Cross Validation

Suppose we are using the **k-Nearest Neighbours (kNN)** algorithm.

The hyperparameter is

\[
K
\]

Different values are tested, such as

\[
K=1,\;3,\;5,\;7,\;9,\;11
\]

For each value of **K**:

1. Perform **K-Fold Cross Validation**.
2. Compute the average validation accuracy.
3. Compare the results.
4. Select the value of **K** that gives the highest average accuracy.

After selecting the best hyperparameter, train the final model using the complete training data and evaluate it once on the **Testing Set**.

---

# Advantages of Cross Validation

- Provides a more reliable estimate of model performance.
- Every sample is used for both training and validation.
- Reduces the chances of overfitting.
- Makes better use of limited data.
- Helps in selecting the best hyperparameters.
- Produces a more stable evaluation than a single train-validation split.

---

# Disadvantages of Cross Validation

- Computationally expensive.
- Requires training the model multiple times.
- Takes more execution time than a simple train-test split.
- May not be suitable for very large datasets.

---

# Difference Between Validation Split and Cross Validation

| Validation Split | K-Fold Cross Validation |
|------------------|-------------------------|
| Uses only one validation set. | Uses multiple validation sets. |
| Faster to execute. | Slower because the model is trained K times. |
| Less reliable. | More reliable and stable. |
| Performance depends on one split. | Performance is averaged over all folds. |

---

# Key Points

- **K-Fold Cross Validation** divides the training dataset into **K equal folds**.
- During each iteration, **one fold is used for validation**, while the remaining **K − 1 folds are used for training**.
- This process continues until **every fold has served as the validation set exactly once**.
- The **average validation accuracy** is used to evaluate the model.
- After selecting the best model, the **Testing Set** is used only once to measure the final performance.

---

# Summary

The complete workflow of K-Fold Cross Validation is:

1. Split the original dataset into **Training Data** and **Testing Data**.
2. Divide the **Training Data** into **K equal folds**.
3. Train the model **K times**, using a different fold for validation in each iteration.
4. Compute the **average validation accuracy** across all folds.
5. Select the best-performing model and hyperparameters.
6. Finally, evaluate the selected model on the **Testing Set** to obtain the final unbiased performance.

> **K-Fold Cross Validation is one of the most widely used techniques in Machine Learning because it provides a reliable estimate of model performance while making efficient use of the available training data.**

````python
# Cross Validation for hyperparam tuning using GridSearchCV

from sklearn.model_selection import GridSearchCV

model = KNeighborsClassifier()
param_grid = {"n_neighbors": [3, 5, 7, 9]}

classifierCV = GridSearchCV(
    model,
    param_grid,
    cv=5,
    scoring="recall"
)

classifierCV.fit(x_train_scaled, y_train)

y_pred = classifierCV.predict(x_test_scaled)

print("recall score: ", recall_score(y_test, y_pred))
print("accuracy score: ", accuracy_score(y_test, y_pred))
print("precision score: ", precision_score(y_test, y_pred))

# results
res = pd.DataFrame(classifierCV.cv_results_)
print(res[["param_n_neighbors", "mean_test_score"]])

print(classifierCV.best_params_)
````

````python
# Pipeline
from sklearn.pipeline import Pipeline

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

pipeline = Pipeline([
    ('scaler', StandardScaler()), 
    ('knn', KNeighborsClassifier())
])

param_grid = {"knn__n_neighbors": [3, 5, 7, 9]}

classifierCV = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="recall"
)

classifierCV.fit(x_train, y_train)

y_pred = classifierCV.predict(x_test)

print("recall score: ", recall_score(y_test, y_pred))
print("accuracy score: ", accuracy_score(y_test, y_pred))
print("precision score: ", precision_score(y_test, y_pred))

print(classifierCV.best_params_)
````
