# Overfitting

## What is Overfitting?

Overfitting occurs when a Machine Learning model becomes **too complex** and learns not only the actual patterns in the training data but also the noise and random fluctuations.

As a result, the model performs very well on the training data but poorly on unseen (test) data.

---

## Example

Training Accuracy:

```text
95%
```

Test Accuracy:

```text
65%
```

This large gap indicates that the model is **Overfitting**.

---

## Characteristics of Overfitting

- Model is too complex.
- Fits the training data almost perfectly.
- Learns noise instead of general patterns.
- Performs poorly on unseen data.
- Has poor generalization ability.

---

## Bias and Variance

### Low Bias

The model makes very few errors on the training data because it closely fits the training examples.

### High Variance

The model is highly sensitive to small changes in the training data.

A slight change in the dataset may produce a very different model.

---

## Graphical Representation

### Ideal Model

The model captures the overall trend of the data.

```text
      *
    *   *
  *       *
 *         *
-------------
```

---

### Overfitted Model

The model tries to pass through every data point.

```text
 ~~~/\_/\/\__/\/\_/~~
```

Instead of learning the actual relationship, it memorizes the training data.

---

## Why Overfitting Happens

- Model is too complex.
- Too many features.
- Training data is too small.
- Too many training iterations.
- Presence of noise in the data.

---

## Problems of Overfitting

- High Training Accuracy
- Low Test Accuracy
- Poor Generalization
- Unstable Predictions
- High Variance

---

## How to Prevent Overfitting

- Collect more training data.
- Remove unnecessary features (Feature Selection).
- Use Cross Validation.
- Apply Regularization (L1 or L2).
- Reduce model complexity.
- Use Early Stopping (for Deep Learning).
- Perform proper Feature Engineering.

---

## Key Points

- Overfitting means the model memorizes the training data instead of learning general patterns.
- Training accuracy is high, but test accuracy is low.
- Overfitting has **Low Bias** and **High Variance**.
- The model becomes sensitive to small changes in the data.
- The goal is to build a model that performs well on both training and test data.

# Underfitting

## What is Underfitting?

Underfitting occurs when a Machine Learning model is **too simple** to learn the underlying patterns in the data.

As a result, the model performs poorly on both the training data and unseen (test) data.

---

## Example

Training Accuracy:

```text
70%
```

Test Accuracy:

```text
65%
```

The model cannot capture the relationship between input and output.

---

## Characteristics of Underfitting

- Model is too simple.
- Fails to learn important patterns.
- High error on training data.
- High error on test data.
- Poor prediction performance.

---

## Bias and Variance

### High Bias

The model makes many incorrect assumptions about the data.

As a result, it cannot learn the actual relationship.

### Low Variance

The model is not very sensitive to changes in the training data.

Even if the training data changes slightly, the model remains almost the same.

---

## Example

Suppose the actual relationship is:

```text
y = x²
```

Data:

| x | y |
|---|---|
| 1 | 1 |
| 2 | 4 |
| 3 | 9 |
| 4 | 16 |
| 5 | 25 |

If we fit a **straight line** instead of a curve, the model cannot capture the true relationship.

This is an example of **Underfitting**.

---

## Graphical Representation

### Actual Relationship

```text
        ●
      ●
    ●
  ●
●
```

### Underfitted Model

```text
──────────────
```

The model is too simple to follow the data pattern.

---

## Why Underfitting Happens

- Model is too simple.
- Too few features.
- Insufficient training.
- Excessive regularization.
- Important information is missing.

---

## Problems of Underfitting

- Low Training Accuracy
- Low Test Accuracy
- High Training Error
- High Test Error
- Poor Generalization

---

## How to Prevent Underfitting

- Increase model complexity.
- Add more useful features.
- Create derived and interaction features.
- Reduce regularization.
- Train the model for more iterations.
- Use a more powerful algorithm.

---

# Optimal Model

An ideal Machine Learning model should have:

- Low Bias
- Low Variance

Such a model learns the true patterns in the data and performs well on both training and test datasets.

---

## Key Points

- Underfitting occurs when the model is **too simple**.
- The model cannot learn the underlying pattern in the data.
- Training accuracy and test accuracy are both low.
- Underfitting has **High Bias** and **Low Variance**.
- The goal is to achieve an **Optimal Model** with **Low Bias** and **Low Variance**.

# Overfitting & Underfitting Fixes

The goal of a Machine Learning model is to achieve **Low Bias** and **Low Variance**.

This balance is known as the **Bias-Variance Tradeoff**.

---

# Bias-Variance Tradeoff

An ideal model should have:

- Low Bias
- Low Variance

If Bias is too high, the model **Underfits**.

If Variance is too high, the model **Overfits**.

The objective is to find the right balance between the two.

---

# Fixing Underfitting

Underfitting occurs because the model is **too simple**.

It has:

- High Bias
- Low Variance

## 1. Increase Model Complexity

Use a more complex model so it can learn the underlying patterns.

Examples:

- Increase the number of parameters.
- Increase the number of layers.
- Increase the number of neurons (Deep Learning).

---

## 2. Reduce Regularization

If regularization is too strong, the model becomes too simple.

Reducing L1 or L2 Regularization allows the model to learn more complex patterns.

---

## 3. Add More Useful Features

Create or select informative features.

Examples:

- Derived Features
- Interaction Features
- Polynomial Features

More useful features help the model capture hidden relationships.

---

## 4. More Data

Increasing the amount of data **usually does not solve Underfitting** because the problem is the model's simplicity, not the dataset size.

---

# Fixing Overfitting

Overfitting occurs because the model is **too complex**.

It has:

- Low Bias
- High Variance

## 1. Reduce Model Complexity

Simplify the model.

Examples:

- Reduce the number of parameters.
- Reduce the number of layers.
- Reduce the number of neurons.

---

## 2. Increase Regularization

Apply stronger regularization.

Examples:

- L1 Regularization (Lasso)
- L2 Regularization (Ridge)

Regularization prevents the model from memorizing the training data.

---

## 3. Feature Selection

Remove unnecessary or irrelevant features.

Keeping only important features reduces overfitting.

---

## 4. Increase Training Data

Providing more diverse training data helps the model generalize better and reduces overfitting.

---

# Summary

| Underfitting | Overfitting |
|--------------|-------------|
| High Bias | High Variance |
| Model is too simple | Model is too complex |
| Increase model complexity | Reduce model complexity |
| Reduce regularization | Increase regularization |
| Add more useful features | Remove unnecessary features |
| More data usually does not help | More data often helps |

---

# Key Points

- **Underfitting = High Bias + Low Variance**
- **Overfitting = Low Bias + High Variance**
- The goal is to achieve **Low Bias + Low Variance**.
- This balance is called the **Bias-Variance Tradeoff**.
- A well-balanced model performs well on both training and test data.
