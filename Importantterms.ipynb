# Overfitting

## What is Overfitting?

Overfitting occurs when a Machine Learning model becomes **too complex** and learns not only the actual patterns in the training data but also the noise and random fluctuations.

As a result, the model performs very well on the training data but poorly on unseen (test) data.

---

## Example

Training Accuracy:

```text
95%
```

Test Accuracy:

```text
65%
```

This large gap indicates that the model is **Overfitting**.

---

## Characteristics of Overfitting

- Model is too complex.
- Fits the training data almost perfectly.
- Learns noise instead of general patterns.
- Performs poorly on unseen data.
- Has poor generalization ability.

---

## Bias and Variance

### Low Bias

The model makes very few errors on the training data because it closely fits the training examples.

### High Variance

The model is highly sensitive to small changes in the training data.

A slight change in the dataset may produce a very different model.

---

## Graphical Representation

### Ideal Model

The model captures the overall trend of the data.

```text
      *
    *   *
  *       *
 *         *
-------------
```

---

### Overfitted Model

The model tries to pass through every data point.

```text
 ~~~/\_/\/\__/\/\_/~~
```

Instead of learning the actual relationship, it memorizes the training data.

---

## Why Overfitting Happens

- Model is too complex.
- Too many features.
- Training data is too small.
- Too many training iterations.
- Presence of noise in the data.

---

## Problems of Overfitting

- High Training Accuracy
- Low Test Accuracy
- Poor Generalization
- Unstable Predictions
- High Variance

---

## How to Prevent Overfitting

- Collect more training data.
- Remove unnecessary features (Feature Selection).
- Use Cross Validation.
- Apply Regularization (L1 or L2).
- Reduce model complexity.
- Use Early Stopping (for Deep Learning).
- Perform proper Feature Engineering.

---

## Key Points

- Overfitting means the model memorizes the training data instead of learning general patterns.
- Training accuracy is high, but test accuracy is low.
- Overfitting has **Low Bias** and **High Variance**.
- The model becomes sensitive to small changes in the data.
- The goal is to build a model that performs well on both training and test data.
