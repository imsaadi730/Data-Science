```python
import findspark
```

```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName("app_101").getOrCreate()
```

```python
titanic=spark.read.csv("train.csv" , inferSchema=True,header=True)
```

```python
titanic.show(5)
```

```python
# Select relevant columns
t1=titanic.select("Survived","Pclass","Sex","Age","Sibsp","Parch","Fare","Embarked")
```

```python
t1.show(5)
```

```python
#Check missing values
#String To numeric Conversion
```

```python
t1.describe().show()
```

```python
# Fill Missing values
from pyspark.ml.feature import Imputer
```

```python
age_imp=Imputer(inputCol="Age", outputCol="Age", strategy="median")
```

```python
t2=age_imp.fit(t1).transform(t1).show()
```

```python
from pyspark.ml.feature import StringIndexer
```

```python
indexer=StringIndexer(inputCol="Embarked", outputCol="Embarked_int")
```

```python
t3=indexer.fit(t2).transform(t2)
```

```python
t2 = age_imp.fit(t1).transform(t1)
t2.show()
```

```python
# This will now work perfectly since t2 is a valid DataFrame!
t3 = indexer.fit(t2).transform(t2)
t3.show(5)
```
